```python
# create a text object to do calculations on
text = """
How is CF treated?
From 1955 until recently CF treatment was aimed only at the disease symptoms and not the underlying cause. This would include removing mucus using physical methods (like beating the patient’s back firmly), antibiotic treatment of infections, taking pancreatic enzymes to support nutrient absorption, and vitamin supplements. Since 1983, lung transplants have also been an option. These days there are therapies that target the cause of the disease.

Modern treatment for CF includes medication that target the specific mutation causing the disease. Because the treatment is specific to the mutation, this is called precision medicine.  Some treatments try to correct a dysfunctioning CFTR, while other try to improve the function of mutated CFTR (Connett, 2019). There are also treatments that try to increase the amount of CFTR produced (Haq et al, 2022). 

The first precision medication developed to treat CF is called ivacaftor. It works by wedging open the mutated CFTR channel, which improves its function. This medication works best in patients that have a mutation affecting CFTR function. As we saw above, most patients have a mutation (F508del) that leads to a destruction of CFTR, and so this treatment does not work very well in these patients (Flume et al, 2012).

The second medication was developed in 2015 and is called orkambi. This treatment combines ivacaftor with a second medication called lumacaftor. Lumacaftor is a corrector that helps to correct the shape of the mutated CFTR protein. This combination is used specifically in patients who have the F508del mutation (Connett, 2019).

In 2018, a third medication called symkevi was created.  It combines ivacaftor with another medication called tezacaftor. Tezacaftor is also a corrector, like lumacaftor. This medication is used in patients with the F508del mutation, but also in patients that have 14 very specific mutations.

Recently the most effective treatment for CF so far was developed. It is called Kaftrio and combines three other medications, namely ivacaftor, tezacaftor and elexacaftor (Figure 2). Elexacaftor is another corrector, but it corrects the CFTR protein at a different spot than tezacaftor, therefore helping the protein even more. Because it is such a powerful medication, it is prescribed for patients with the F508del mutation or any other mutation. 

"""
```

```python
# Do an initial exploration of the text, in terms of total words and frequencies.
# import modules to manipulate text
import re
from collections import Counter

# Extract words, including apostrophes and hyphens
words = re.findall(r"\b[\w]+(?:['’\-][\w]+)*\b", text.lower())

print("Total words:", len(words))
print("Unique words:", len(set(words)))

# Top 10 most frequent words
frequencies = Counter(words)
print(frequencies.most_common(10))
```

## Calculate the probability of the word "medication" appearing.  
Underlying probability equation: P(E) = count(E)/n where E is the event we are looking for and n is the total number of experiments/observations (number of words in this case).

```python
# P(E) where E is the word "medication"
# note that this excludes variations of the word, like the plural "medications" 
medication = len(re.findall("medication", text.lower()))
total_words = len(words)

# manually calculate the probability by dividing the count by the total number
prob_E = medication/total_words
print(prob_E)
```

The probability of seeing the word "medication" in the text is 0.027 (2.7%).  

## Calculate the probability of the word "medication" conditional on the previous word being "this"  
Underlying probability equation: P(E|F) = [count(E) and count(F)]/count(F), where F is the observation of the word "this". We do this because we are subsetting the data to only words that are preceded by "this". 

```python
this = len(re.findall("this", text.lower()))

prob_F = this/total_words
print(prob_F)
```

The probability of seeing the word "this" in the text is 0.019 (1.9%). We can do a simple count in the text to identify the proportion of times that "this" is followed by "medication", when we already know the first word is "this".

```python
this_medication_count = 0

for i in range(len(words) - 1):
    if words[i] == "this" and words[i + 1] == "medication":
        this_medication_count += 1

print(this_medication_count)

probability = this_medication_count / this

print(probability)
```

We have calculated that the probability of "this" being followed by "medication" is 0.286 (28.6%). However, we have to bear in mind that our calculations are made on an extremely small data set. How confident can we be in this result? To quantify this, we can calculate the 95% confidence interval as 1/sqrt(n), where n = the total number of times "this" appears.

```python
# import module to calculate square root
import math

# define number of trials
n = this

# calculate confidence interval
conf_int_95 = 1/math.sqrt(n)

print(conf_int_95)
```

The confidence interval is 0.38, meaning that the true value for "this" being followed by "medication" is the calculated value +/- the 96% confidence interval, so 0.286 +/- 0.378, so our confidence in the result isn't very high, and we'll definitely need a larger text sample to do this calculation. 

This conditional probabilistic analysis of text is the underpinning principle of large language models (LLMS), except that they don't just take one previous word into consideration, but whole strings of text. This causes a large computational burden on the model, which is why there is a rapid increase in data centres. In addition, the predicted word is influenced by the text on which the model is based and the prefernces of the person who trained the model.
